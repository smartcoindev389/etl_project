```python
import pandas as pd
import numpy as np
from pathlib import Path
from google.colab import drive

# Solución 1: Instalar dependencias automáticamente
try:
    from unidecode import unidecode
except ImportError:
    print("Instalando unidecode...")
    !pip install unidecode
    from unidecode import unidecode


def agregar_NPS():
    """
    Función que agrega campos de NPS (Net Promoter Score) al archivo principal
    mediante un LEFT JOIN para mantener todos los registros originales.
    """
    try:
        print("🔍 Iniciando proceso de agregación de NPS...")



        # 2. Definir rutas de archivos
        ruta_base = Path("/content/drive/My Drive/ProyectoSpeachAnalytics/FOH/TranscripcionINBOUND/resultado_metadata_flagsSA")

        # Verificar existencia de archivos
        archivos_requeridos = {
            'principal': ruta_base / "resultados_analisis_completo_metadata_final.csv",
            'NPS': ruta_base / "BASE_NPS.xlsx",
            'salida': ruta_base / "resultados_analisis_completo_metadata_final_nps.csv"
        }

        for nombre, ruta in archivos_requeridos.items():
            if not ruta.exists() and nombre != 'salida':
                raise FileNotFoundError(f"🚨 No se encontró el archivo: {ruta}")

        # 3. Cargar archivo principal
        print("\n📥 Cargando tabla principal...")
        df_principal = pd.read_csv(archivos_requeridos['principal'], low_memory=False)
        print(f"✅ Registros cargados: {len(df_principal):,}")

        # 4. Cargar y preparar datos de NPS
        print("\n📊 Procesando base de NPS...")
        df_nps = pd.read_excel(archivos_requeridos['NPS'])

        # Limpieza y preparación de datos
        df_nps = df_nps.rename(columns={'CONVERSATION ID': 'conversation_id'})
        columnas_necesarias = ['conversation_id', 'PREGUNTA', 'RESPUESTA']
        df_nps = df_nps[columnas_necesarias].drop_duplicates('conversation_id')
        print(f"✅ Registros únicos en NPS: {len(df_nps):,}")

        # 5. Unir los datos (LEFT JOIN)
        print("\n🔗 Realizando fusión de datos...")
        df_final = pd.merge(
            df_principal,
            df_nps,
            on='conversation_id',
            how='left'
        )

        # 6. Procesamiento de texto mejorado
        print("\n🔠 Procesando texto...")

        # Función segura para unidecode
        def safe_unidecode(x):
            if pd.isna(x):
                return np.nan
            try:
                return unidecode(str(x))
            except:
                return str(x)

        df_final['PREGUNTA'] = df_final['PREGUNTA'].apply(safe_unidecode)

        print(df_final['PREGUNTA'].value_counts(dropna=False))

        # 6. Procesar y crear flags
        print("\n🏷️ Creando variables adicionales...")
        df_final['RESPUESTA'] = pd.to_numeric(df_final['RESPUESTA'], errors='coerce').fillna(-1)

        df_final['FLG_BUENO'] = np.where(
            df_final['RESPUESTA'] == -1,
            -1,
            np.where(df_final['RESPUESTA'] > 7, 1, 0)
        )

        # 7. Guardar resultados
        print("\n💾 Guardando archivo final...")
        df_final.to_csv(archivos_requeridos['salida'], index=False, encoding='utf-8-sig')

        print("\n✅✅✅ Proceso completado exitosamente!")
        print(f"📂 Archivo guardado en: {archivos_requeridos['salida']}")
        print(f"📊 Total de registros: {len(df_final):,}")

        # Mostrar resumen de resultados
        print("\n📈 Resumen de NPS:")
        print(df_final['RESPUESTA'].value_counts(dropna=False).sort_index())
        print("\n🏷️ Resumen de FLG_BUENO:")
        print(df_final['FLG_BUENO'].value_counts(dropna=False))

    except Exception as e:
        print(f"\n❌❌❌ Error crítico: {str(e)}")
        raise

if __name__ == "__main__":
    agregar_NPS()
```
